```python
import sys

# --- Custom Data Structure: Min Heap ---
class MinHeap:
    def __init__(self):
        self.heap = []

    def push(self, priority, item):
        """Push an item with a priority onto the heap."""
        self.heap.append((priority, item))
        self._sift_up(len(self.heap) - 1)

    def pop(self):
        """Pop the item with the lowest priority."""
        if not self.heap:
            return None
        if len(self.heap) == 1:
            return self.heap.pop()

        root = self.heap[0]
        # Move last element to root
        self.heap[0] = self.heap.pop()
        self._sift_down(0)
        return root

    def is_empty(self):
        return len(self.heap) == 0

    def _sift_up(self, idx):
        parent = (idx - 1) // 2
        while idx > 0 and self.heap[idx][0] < self.heap[parent][0]:
            self.heap[idx], self.heap[parent] = self.heap[parent], self.heap[idx]
            idx = parent
            parent = (idx - 1) // 2

    def _sift_down(self, idx):
        smallest = idx
        left = 2 * idx + 1
        right = 2 * idx + 2
        n = len(self.heap)

        if left < n and self.heap[left][0] < self.heap[smallest][0]:
            smallest = left
        if right < n and self.heap[right][0] < self.heap[smallest][0]:
            smallest = right

        if smallest != idx:
            self.heap[idx], self.heap[smallest] = self.heap[smallest], self.heap[idx]
            self._sift_down(smallest)
```

```python
# --- Data Structure: Graph ---
class Graph:
    def __init__(self):
        self.adjacency_list = {}

    def add_node(self, node):
        if node not in self.adjacency_list:
            self.adjacency_list[node] = []

    def add_edge(self, u, v, weight):
        # Undirected graph
        self.adjacency_list[u].append((v, weight))
        self.adjacency_list[v].append((u, weight))

    def get_nodes(self):
        return list(self.adjacency_list.keys())
```

```python
# --- Algorithm: Dijkstra ---
def dijkstra(graph, start, end):
    distances = {node: float('infinity') for node in graph.get_nodes()}
    previous_nodes = {node: None for node in graph.get_nodes()}
    distances[start] = 0

    priority_queue = MinHeap()
    priority_queue.push(0, start)

    while not priority_queue.is_empty():
        current_dist, current_node = priority_queue.pop()

        # If we found a shorter path to current_node previously, skip
        if current_dist > distances[current_node]:
            continue

        if current_node == end:
            break

        for neighbor, weight in graph.adjacency_list[current_node]:
            dist = current_dist + weight
            if dist < distances[neighbor]:
                distances[neighbor] = dist
                previous_nodes[neighbor] = current_node
                priority_queue.push(dist, neighbor)

    return distances, previous_nodes
```

```python
# --- Helper to reconstruct path ---
def get_shortest_path(previous_nodes, start, end):
    path = []
    current = end
    while current is not None:
        path.append(current)
        current = previous_nodes[current]
    path.reverse()
    if path[0] == start:
        return path
    return None # No path found
```

```python
g = Graph()

# Now the graph knows about these specific cities
nodes = ['Mysore', 'Bangalore', 'Mandya', 'Channapatna', 'Ramanagara', 'Kengeri']
for node in nodes:
    g.add_node(node)

# Distances are hypothetical examples
edges = [
    ('Mysore', 'Mandya', 40),
    ('Mandya', 'Channapatna', 45),
    ('Channapatna', 'Ramanagara', 30),
    ('Ramanagara', 'Kengeri', 40),
    ('Kengeri', 'Bangalore', 15),
    ('Mandya', 'Bangalore', 100) # A hypothetical longer direct route
]

for u, v, w in edges:
    g.add_edge(u, v, w);

print("--- Smart City Route Navigator ---")
print("Available Locations:", nodes)

# Input handling with validation
while True:
    start_node = input("Enter Start Location: ").title()
    if start_node in nodes:
        break
    else:
        print(f"Invalid start location: '{start_node}'. Please choose from {nodes}")

while True:
    end_node = input("Enter Destination: ").title()
    if end_node in nodes:
        break
    else:
        print(f"Invalid destination: '{end_node}'. Please choose from {nodes}")
```

```python
distances, previous_nodes = dijkstra(g, start_node, end_node)

if distances[end_node] == float('infinity'):
    print(f"No path found from {start_node} to {end_node}")
else:
    path = get_shortest_path(previous_nodes, start_node, end_node)
    print(f"Shortest path from {start_node} to {end_node}: {path}")
    print(f"Total distance: {distances[end_node]} KM")
```
